# Nemotron LoRA Runner
This notebook delegates execution to `nemotron.py` so the Kaggle notebook stays small and the script remains the single source of truth.

If Kaggle assigns a P100, the script will stop early with a clear message telling you to restart the session until you get a newer GPU.


```python
import os

ACTIVE_VARIANT = os.environ.get("NEMOTRON_VARIANT", "competition_plus_gsm8k")
print(f"Using variant: {ACTIVE_VARIANT}")
print("Set os.environ['NEMOTRON_VARIANT'] before running the next cell to switch variants.")
print("Available variants: smoke_debug, competition_plus_gsm8k, higher_rank")
```

```python
import runpy
from pathlib import Path

script_path = Path("nemotron.py")
if not script_path.exists():
    raise FileNotFoundError(f"Missing runner script: {script_path.resolve()}")

print(f"Running {script_path.resolve()}")
runpy.run_path(str(script_path), run_name="__main__")
```
